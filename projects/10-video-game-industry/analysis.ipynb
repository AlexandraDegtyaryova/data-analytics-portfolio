 **Числовые значения с плавающей запятой (float64).**

- `na_sales` — хранит продажи в Северной Америке (в миллионах проданных копий), тип верный, но можно уменьшить разрядность до **float16**
- `other_sales` — содержит продажи в других странах (в миллионах проданных копий), тип верный но можно уменьшить разрядность до **float16**
- `critic_score` — оценка критиков (от 0 до 100), содержит только целые значения, тип стоит поменять на **int8** (с учётом большого количесва пропусков)


**Строковые данные (object).** Восемь столбцов имеют тип данных `object`:
- `name` — название игры, содержит стороковую информацию, верный тип данных 
- `platform` — название платформы, можно поменять object на **category**
- `genre` — жанр игры, можно поменять object на **category**
- `eu_sales` — продажи в Европе (в миллионах проданных копий), содержит числовую дробную информацию, надо поменять object на **float16**
- `jp_sales` — продажи в Японии (в миллионах проданных копий), содержит числовую дробную информацию, надо поменять object на **float16**
- `user_score` — оценка пользователей (от 0 до 10), поменять на **float16**
- `rating` — рейтинг организации ESRB (англ. Entertainment Software Rating Board), строковая информация, можно поменять object на **category**

```python
# меняем тип данных с учетом пропусков и памяти
numeric_conversions = {
    'year_of_release': 'Int16',
    'critic_score': 'Int8', 
    'user_score': 'float16',
    'eu_sales': 'float16',
    'jp_sales': 'float16'
}

for column, dtype in numeric_conversions.items():
    df[column] = pd.to_numeric(df[column], errors='coerce').astype(dtype)
    
```

```python
# изменение типа данных для оптимизации использования памяти
category_columns_float = ['na_sales','other_sales']
df[category_columns_float] = df[category_columns_float].astype('float16') 

category_columns_ctg = ['platform', 'genre']
df[category_columns_ctg] = df[category_columns_ctg].astype('category')
```

```python
# проверяем 
df.info()
```

### Наличие пропусков в данных

- Посчитайте количество пропусков в каждом столбце в абсолютных и относительных значениях.


```python
# выводим количество пропущенных строк в датафрейме
df.isna().sum()
```

```python
# подсчитываем процент строк с пропусками
df.isna().sum() / len(df) * 100
```

**Пропуски содержат:**
- `name` - всего 2 значения, очень маленький процент, можно удалить или оставить как есть
- `year_of_release` (пропуски в виде NaN) 275 пропусков, почти 2%  маленькое количество пропусков, оставить как/заменить на ноль/удалить
- `genre` тоже всего 2 пропуска, оставить как есть или удалить
- `eu_sales` и `jp_sales` маленькое количество пропусков, можно заменить средним
- `critic_score` и `user_score` (пропуски в виде NaN) больше половины данных - пропуски, заменить на значение-индикатор
- `rating` тоже много пропусков (40%) можно заполнить в зависимости от кол-ва продаж

У `name` и `genre` пропуски это скорее всего случайные ошибки в данных, у `year_of_release` причина появления пропусков пока не особо ясна (вероятно дубликаты или ошибки сбора данных). `critic_score` , `user_score` , `rating` имеют пропуски из-за особенности сбора данных тк не каждый фильм имеет оценку, например, в силу года. У слишком старых фильмов не скорее всего не будет оценок.

```python
# удаляем строки с маленьким % пропусков 

df = df[~df['name'].isna()]
df = df[~df['genre'].isna()]
df = df[~df['eu_sales'].isna()]
df = df[~df['jp_sales'].isna()]
df = df[~df['year_of_release'].isna()]

# заменяем пустые значения в critic_score и user_score индикатором -1
df['critic_score'] =df['critic_score'].fillna(-1)
df['user_score'] =df['user_score'].fillna(-1)

# (рейтинг в данном случае не может быть нулевым, нулём обозначаем пропуск)
df['rating'] =df['rating'].fillna(0)

# меняем тип у rating
df['rating'] = df['rating'].astype('category')
```

```python
# проверяем результат обработки пропусков
df.isna().sum()
```

##  Категоризация данных
    
Проведите категоризацию данных:
- Разделите все игры по оценкам пользователей и выделите такие категории: высокая оценка (от 8 до 10 включительно), средняя оценка (от 3 до 8, не включая правую границу интервала) и низкая оценка (от 0 до 3, не включая правую границу интервала).

```python
df_actual['user_score_level'] = pd.cut(df_actual['user_score'], bins=[0, 3, 8, 11], labels=["low", "medium", "high"],right=False)
```

- Разделите все игры по оценкам критиков и выделите такие категории: высокая оценка (от 80 до 100 включительно), средняя оценка (от 30 до 80, не включая правую границу интервала) и низкая оценка (от 0 до 30, не включая правую границу интервала).

```python
df_actual['critic_score_level'] = pd.cut(df_actual['critic_score'], bins=[0, 30, 80, 101], labels=["low", "medium", "high"],right=False)
```

```python
# проверяем 
df_actual.head(4)
```

- После категоризации данных проверьте результат: сгруппируйте данные по выделенным категориям и посчитайте количество игр в каждой категории.

```python
df_actual.groupby(['user_score_level','critic_score_level'])['name'].count()
```

- Выделите топ-7 платформ по количеству игр, выпущенных за весь актуальный период.

```python
# Группируем по платформе, считаем количество игр и сортируем по убыванию
platform_games = df_actual.groupby('platform')['name'].count().sort_values(ascending=False)

# Выводим первые 7 платформ
platform_games.head(7)
```
